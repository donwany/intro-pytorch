```python
import torch
```

```python
x = torch.tensor(2.0)
y = torch.tensor(4.0)

# y = m * x

m = torch.tensor(0.5, requires_grad=True)

learning_rate = 0.1 # often people use 0.01, 0.001, 0.0001
epochs = 5 # how many times we want to loop through our data

for epoch in range(epochs):

    # 1. Forward pass
    prediction = m * x # 0.5 * 2.0 = 1.0

    # 2. Loss
    loss = (prediction - y) ** 2 # (1.0 - 4.0) ** 2 = 9.0

    # 3. Backward pass (calculate gradients)
    loss.backward()

    print("Gradient:", m.grad)

    # 4. Update weights
    with torch.no_grad():
        m -= learning_rate * m.grad # m = m - learning_rate * m.grad

    # 5. Zero gradients
    m.grad.zero_() # m.grad = 0

    print("Updated weight:", m.item())
```
